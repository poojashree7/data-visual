```python
import matplotlib.pyplot as plt
x=[101,102,103,104,105]
y=[80,90,70,65,89]
plt.xlabel("Roll Number")
plt.ylabel("Marks")
plt.xlim(101,106)
plt.ylim(65,95)
plt.title("Student Performance")
plt.plot(x,y,color='pink',linestyle='dashdot',linewidth=3,marker='*',markerfacecolor='black',markersize=12)
plt.savefig("sample1.png")
plt.show()
```

```python
import matplotlib.pyplot as plt
students=["A","B","C","D","E"]
marks=[60,45,70,50,80]
plt.bar(students,marks,color="pink",width=0.7,edgecolor="black")
plt.title("student marks in semester")
plt.xlabel("students")
plt.ylabel("marks")
plt.show()
```

```python
hours=[2,3,4,5,6]
marks=[40,55,60,45,65]
plt.scatter(hours,marks,color="black",marker="^",s=120,label="Studentsmarks")
plt.legend()
```

```python
import pandas as pd
mk=pd.Series(marks)
mk
```

```python
mk.plot(kind="density",color="skyblue",linewidth=2,linestyle="--",label="marks")
plt.legend()
plt.grid(True)
plt.show()
```
