```python
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv("C:/Users/Ibscai22/iris.csv")
x=df["sepal_length"]
y=df["sepal_width"]
a=df["petal_length"]
b=df["petal_width"]
plt.xlabel("Sepal")
plt.ylabel("Petal")
plt.title("Line Plot Example")
plt.plot(x,y,label="Line-1",color='skyblue',linestyle='dashed',linewidth=3,marker='.',markerfacecolor='purple',markersize=10)
plt.plot(a,b,label="Line-2",color='pink',linestyle='dashed',linewidth=3,marker='*',markerfacecolor='black',markersize=10)
plt.legend()
plt.savefig("iris.png")
plt.show()
```
